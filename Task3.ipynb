# Housing Price Prediction

## Project Overview

The objective of this project is to build a Machine Learning model that predicts housing prices using the California Housing dataset from Kaggle.

The project involves data loading, data cleaning, feature selection, model training, and performance evaluation. Features such as location, house age, total rooms, population, median income, and ocean proximity will be used to predict median house values.

## Objectives

- Load and explore the housing dataset.
- Rename columns for better readability.
- Identify and handle missing values.
- Select relevant features for prediction.
- Train a Machine Learning regression model.
- Evaluate model performance using appropriate metrics.

## Tools and Libraries

Python, Pandas, NumPy, Matplotlib, Seaborn, and Scikit-learn.

## Target Variable

`House_Value` — the median house value that the model will learn to predict.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```
